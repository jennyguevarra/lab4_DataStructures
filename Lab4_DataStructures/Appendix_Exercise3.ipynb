```python
SEED_NUM = 2  
STUDENT_MAJOR = "BSECE"
FAVORITE_ARTIST = "Ben&Ben"
LOCAL_HAZARD = "FLOOD"

CONTROL_NUM = max(3, SEED_NUM)
```

```python
from collections import deque, Counter

# initialize deque with maxlen = 5
dq = deque(maxlen=5)

# initial burst
burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

# Data Expansion (add exactly CONTROL_NUM + 2 alerts)
for i in range(CONTROL_NUM + 2):
    burst.append((CONTROL_NUM + 7 + i, "WEATHER"))

# trackers
weather_check = False
hazard_check = False

# process burst
for timestamp, category in burst:
    # get current categories in deque
    current_categories = [c for _, c in dq]
    count = Counter(current_categories)

    # Spam Rule: reject if category appears 2 or more times
    if count[category] >= 2:
        continue
    else:
        dq.append((timestamp, category))

    # check conditions
    if timestamp == CONTROL_NUM + 3 and category == "WEATHER":
        weather_check = True

    if category == LOCAL_HAZARD:
        hazard_check = True

# compute sum of timestamps
final_sum = sum(t for t, _ in dq)

# OUTPUTS (ito yung kailangan sa assessment table)
print("LOCAL_HAZARD used:", LOCAL_HAZARD)
print("Total Burst Size:", len(burst))
print("Did WEATHER at CONTROL_NUM + 3 append?:", weather_check)
print("Did LOCAL_HAZARD append?:", hazard_check)
print("Final Deque Length:", len(dq))
print("Final Sum of Timestamps:", final_sum)
print("Exact Final Deque Output:", list(dq))
```
