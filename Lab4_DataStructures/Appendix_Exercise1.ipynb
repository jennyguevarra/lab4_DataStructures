```python
SEED_NUM = 2  
STUDENT_MAJOR = "BSECE"
FAVORITE_ARTIST = "Ben&Ben"
LOCAL_HAZARD = "FLOOD"

CONTROL_NUM = max(3, SEED_NUM)
```

```python
from collections import defaultdict, Counter

# 1. Initialize a simulated stream of incoming applications
stream = [
    ("TUP MANILA", "BSECE"),
    ("TUP TAGUIG", f"BSME_V{CONTROL_NUM}"),
    ("TUP MANILA", f"BSME_V{CONTROL_NUM}"),
    ("TUP MANILA", "BSECE"),
    ("TUP VISAYAS", STUDENT_MAJOR),
    ("TUP TAGUIG", "BSECE"),
    ("TUP MANILA", "BSECE")
]

# 2. Data Expansion: Add exactly (CONTROL_NUM + 3) custom application tuples
for i in range(CONTROL_NUM + 3):
    # Nagdadagdag tayo ng custom programs sa Manila para sa expansion
    stream.append(("TUP MANILA", f"CUSTOM_PROG_{i}"))

# 3. Algorithm Phase 1 (Aggregation): Use defaultdict(list)
campus_dict = defaultdict(list)
for campus, program in stream:
    campus_dict[campus].append(program)

# 4. Algorithm Phase 2 (Frequency Analysis): Use Counter
# Gagawa tayo ng counter para sa bawat campus sa dictionary
analysis_results = {}
for campus, programs in campus_dict.items():
    analysis_results[campus] = Counter(programs)

# 5. Extract the top requested program for Manila campus
manila_counter = analysis_results["TUP MANILA"]
top_program_tuple = manila_counter.most_common(1)[0] # Returns (Program, Frequency)

# --- OUTPUTS FOR ASSESSMENT DATA TABLE ---
print("--- EXERCISE 1 ASSESSMENT DATA ---")
print(f"CONTROL_NUM Used: {CONTROL_NUM}")
print(f"STUDENT_MAJOR Used: {STUDENT_MAJOR}")
print(f"Expanded Stream Length: {len(stream)}")
print(f"Total Applications Processed for TUP Manila: {len(campus_dict['TUP MANILA'])}")
print(f"Top Requested Program at TUP Manila: {top_program_tuple[0]}")
print(f"Exact Frequency of Top Program: {top_program_tuple[1]}")
```
