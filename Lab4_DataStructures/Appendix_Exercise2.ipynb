```python
SEED_NUM = 2  
STUDENT_MAJOR = "BSECE"
FAVORITE_ARTIST = "Ben&Ben"
LOCAL_HAZARD = "FLOOD"

CONTROL_NUM = max(3, SEED_NUM)
```

```python
# Festival set
festival = {
    "Ben&Ben", "SB19", "Bini", "Eraserheads",
    FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"
}

# Users
user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

# Data Expansion (add CONTROL_NUM artists)
festival.update({f"ExtraArtist{i}" for i in range(CONTROL_NUM)})

# Jaccard Function
def jaccard(A, B):
    return (len(A & B) / len(A | B)) * 100

# Calculations
intersection_a = len(user_a & festival)
union_a = len(user_a | festival)
similarity_a = jaccard(user_a, festival)

intersection_b = len(user_b & festival)
union_b = len(user_b | festival)
similarity_b = jaccard(user_b, festival)

dealbreakers = user_a - festival

# Outputs
print("Total Artists in Festival:", len(festival))
print("Intersection (User A):", intersection_a)
print("Union (User A):", union_a)
print("User A Similarity:", similarity_a)

print("User B Similarity:", similarity_b)
print("User A Dealbreakers:", dealbreakers)
```
